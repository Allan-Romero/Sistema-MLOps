# Limpieza de datos — S1-J4

En esta sección se realiza la limpieza inicial del dataset Telco Customer Churn.

Se revisarán:

- Registros duplicados.
- Columnas que no aportan información predictiva.
- Valores nulos.
- Valores vacíos o problemáticos.
- Tipos de datos inconsistentes.

El objetivo es obtener un conjunto de datos limpio y consistente para las siguientes etapas del modelado.

```python
import pandas as pd

df = pd.read_csv("../data/raw/churn.csv")

print("Dimensión original:", df.shape)
```

```python
duplicados = df.duplicated().sum()

print("Registros duplicados:", duplicados)
```

```python
print("Valores únicos customerID:", df["customerID"].nunique())
print("Total de registros:", len(df))
```

```python
print("Tipo de dato de TotalCharges:", df["TotalCharges"].dtype)

valores_vacios = df["TotalCharges"].astype(str).str.strip().eq("").sum()

print("Valores vacíos en TotalCharges:", valores_vacios)
```

```python
filas_totalcharges_vacio = df[
    df["TotalCharges"].astype(str).str.strip().eq("")
]

print("Cantidad de filas problemáticas:", len(filas_totalcharges_vacio))

filas_totalcharges_vacio[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

## Tratamiento aplicado

Durante la revisión se identificaron 11 valores vacíos en la variable `TotalCharges`.

Los 11 registros presentan `tenure = 0`, lo que indica que corresponden a clientes sin antigüedad acumulada. Por esta razón, se decidió conservar los registros y reemplazar los valores vacíos de `TotalCharges` por `0`.

Posteriormente, `TotalCharges` se convierte a formato numérico.

También se elimina la columna `customerID` de la copia utilizada para modelado, debido a que contiene un identificador único para cada uno de los 7043 registros y no representa una característica predictiva del cliente.

No se encontraron registros duplicados.

```python
df_limpio = df.copy()

print("Copia del dataset creada correctamente.")
print("Dimensión inicial:", df_limpio.shape)
```

```python
df_limpio["TotalCharges"] = (
    df_limpio["TotalCharges"]
    .astype(str)
    .str.strip()
    .replace("", "0")
)

df_limpio["TotalCharges"] = pd.to_numeric(
    df_limpio["TotalCharges"],
    errors="coerce"
)

print("Tipo de TotalCharges:", df_limpio["TotalCharges"].dtype)
print("Nulos en TotalCharges:", df_limpio["TotalCharges"].isnull().sum())
```

```python
df_limpio = df_limpio.drop(columns=["customerID"])

print("customerID existe:", "customerID" in df_limpio.columns)
print("Dimensión después de eliminar customerID:", df_limpio.shape)
```

```python
print("========== ANTES DE LA LIMPIEZA ==========")
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])
print("Duplicados reales en dataset original:", df.duplicated().sum())
print("customerID duplicados:", df["customerID"].duplicated().sum())
print("TotalCharges vacíos:", df["TotalCharges"].astype(str).str.strip().eq("").sum())
print("Tipo TotalCharges:", df["TotalCharges"].dtype)
print("customerID presente:", "customerID" in df.columns)

print("\n========== DESPUÉS DE LA LIMPIEZA ==========")
print("Filas:", df_limpio.shape[0])
print("Columnas:", df_limpio.shape[1])
print("Nulos totales:", df_limpio.isnull().sum().sum())
print("Nulos TotalCharges:", df_limpio["TotalCharges"].isnull().sum())
print("Tipo TotalCharges:", df_limpio["TotalCharges"].dtype)
print("customerID presente:", "customerID" in df_limpio.columns)

print(
    "Filas coincidentes después de eliminar customerID:",
    df_limpio.duplicated().sum()
)
```

### Validación de registros coincidentes

El dataset original no contiene registros completamente duplicados y cada valor de `customerID` identifica de manera única a un cliente.

Después de eliminar `customerID`, se identificaron 22 filas coincidentes en las variables restantes. Estas filas no se eliminan, ya que corresponden a clientes distintos que presentan las mismas características y no a duplicados reales del dataset original.

Por lo tanto, se conservan los 7043 registros originales.

```python
columnas_categoricas = df_limpio.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Columnas categóricas:")

for columna in columnas_categoricas:
    print("-", columna)

print("\nCantidad:", len(columnas_categoricas))
```

```python
df_codificado = df_limpio.copy()

df_codificado["Churn"] = df_codificado["Churn"].map({
    "No": 0,
    "Yes": 1
})

print("Variable objetivo codificada:")
print(df_codificado["Churn"].value_counts())
```

```python
categoricas_predictoras = df_codificado.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Variables categóricas predictoras:")
for columna in categoricas_predictoras:
    print("-", columna)

print("\nCantidad:", len(categoricas_predictoras))
```

```python
df_codificado = pd.get_dummies(
    df_codificado,
    columns=categoricas_predictoras,
    dtype=int
)

print("Codificación realizada correctamente.")
print("Dimensión antes:", df_limpio.shape)
print("Dimensión después:", df_codificado.shape)
```

```python
columnas_no_numericas = df_codificado.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("Columnas no numéricas restantes:", columnas_no_numericas)
print("Cantidad:", len(columnas_no_numericas))
```

## Codificación de variables categóricas — S1-J5

Se realizó la transformación de las variables categóricas del dataset para permitir su utilización en algoritmos de machine learning.

### Variable objetivo

La variable `Churn` fue codificada de forma binaria:

- `No` = 0
- `Yes` = 1

### Variables predictoras

Las variables categóricas predictoras fueron transformadas mediante One-Hot Encoding.

Este método crea variables binarias para representar cada categoría sin asignar relaciones ordinales artificiales entre ellas.

Después de la codificación, todas las variables del dataset quedaron representadas numéricamente y listas para las siguientes etapas de preparación y entrenamiento del modelo.
