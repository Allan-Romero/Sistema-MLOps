# Análisis exploratorio inicial — Telco Customer Churn

Este notebook realiza la exploración inicial del dataset utilizado en el proyecto Sistema-MLOps.

El objetivo es identificar la estructura general de los datos, cantidad de registros y columnas, tipos de datos, valores nulos y distribución de la variable objetivo `Churn`.

```python
import pandas as pd

df = pd.read_csv("../data/raw/churn.csv")

print("Dataset cargado correctamente")
print("Filas y columnas:", df.shape)
```

```python
print("Número de filas:", df.shape[0])
print("Número de columnas:", df.shape[1])
```

```python
df.head()
```

```python
print("Tipos de datos:")
print(df.dtypes)
```

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
```

```python
print("Distribución de Churn:")
print(df["Churn"].value_counts())
```

```python
print(df.isnull().sum())
```

```python
print("Distribución porcentual de Churn:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))
```

## Conclusiones de la exploración inicial

- El dataset contiene 7043 registros y 21 columnas.
- La variable objetivo es `Churn`.
- La variable objetivo contiene las categorías `No` y `Yes`.
- La revisión inicial con Pandas no identificó valores nulos en las columnas.
- La distribución de la variable objetivo corresponde a 5174 registros con `Churn = No` y 1869 registros con `Churn = Yes`.
- Se identificaron los tipos de datos y la estructura general del dataset.
- Los resultados de esta exploración servirán como base para las siguientes etapas de limpieza y preparación de los datos.
